## Calcolare il Massimo Comune Divisore

Nell'overview del capitolo abbiamo incontrato il trucco di Euclide: dividere, tenere il resto e ripetere, finché il resto non diventa zero. Qui lo traduciamo in Python e, per metterlo alla prova, lo applichiamo a una coppia di numeri nuova, $60$ e $48$.

```python
# Definizione della funzione: riceve in input due interi positivi

def mcd(a, b):
    while b:
        a, b = b, a % b
    return a

# Esempio di utilizzo

print(mcd(60, 48))
```

Leggiamo il codice riga per riga, come una ricetta:

- `def mcd(a, b):` dà un nome alla ricetta e ai suoi due ingredienti, i numeri `a` e `b`;
- `while b:` significa «ripeti finché `b` non è diventato zero»;
- `a % b` è il simbolo Python per il **resto** della divisione tra `a` e `b`: proprio il resto del trucco di Euclide;
- `a, b = b, a % b` fa scorrere i numeri di un posto: il divisore prende il posto del numero grande, e il resto prende il posto del divisore;
- `return a` consegna il risultato: quando il resto è zero, l'ultimo divisore usato è il MCD.

Seguiamolo con $60$ e $48$: $60$ diviso $48$ dà resto $12$; poi $48$ diviso $12$ dà resto $0$. Resto zero: l'ultimo divisore è $12$, ed è proprio il numero stampato qui sopra. Con il $12$ e l'$8$ dell'overview i passaggi sarebbero gli stessi: resto $4$, poi resto $0$, risultato $4$.

Questo semplice algoritmo, noto come algoritmo di Euclide, è uno dei primi esempi di come un problema complesso possa essere risolto in modo efficiente con un approccio intelligente.
