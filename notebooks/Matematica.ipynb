### In pratica, con NumPy


````python
import numpy as npdef entropia(p):    p = np.asarray(p, dtype=float)    p = p[p > 0]                        # convenzione: 0·log 0 = 0    return -(p * np.log2(p)).sum()equa     = [0.5, 0.5]truccata = [0.9, 0.1]dado     = np.full(6, 1/6)print(entropia(equa))                   # 1.0print(entropia(truccata))               # ~0.4690print(entropia(dado))                   # ~2.5850def cross_entropia(p, q):    p, q = np.asarray(p, dtype=float), np.asarray(q, dtype=float)    m = p > 0    return -(p[m] * np.log2(q[m])).sum()# la realta' e' truccata, il modello crede la moneta equaH_pq = cross_entropia(truccata, equa)   # 1.0kl   = H_pq - entropia(truccata)        # ~0.5310: lo "spreco" in bitprint(H_pq, kl)# perplessita': 2^H, il numero di alternative equiprobabiliprint(2**entropia(equa), 2**entropia(dado))   # 2.0  6.0
````
